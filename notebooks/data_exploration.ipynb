```python
import pandas as pd 

same_convs = pd.read_csv('../data/generative/conversations/NON_ADVERSARIAL_CONVERSATIONS.csv')
adv_convs = pd.read_csv('../data/generative/conversations/ADVERSARIAL_CONVERSATIONS.csv')
```

```python
# Renaming columns
same_convs.rename(columns={'model': 'attacker_model'}, inplace=True)
same_convs.insert(2, 'target_model', same_convs['attacker_model'])
same_convs.info()
```

```python
# Rewrite aversarial conversations ids
starting_index = len(same_convs)
adv_convs['session_id'] = [f'session_{i + starting_index}' for i in adv_convs.index]
```

```python
adv_convs.iloc[-1]
```

```python
merged_df = pd.concat(
    [same_convs, adv_convs],
    ignore_index=True
)

merged_df.info()
```

```python
merged_df.to_csv('../data/generative/conversations/CONVERSATIONS.csv', index=False)
```
